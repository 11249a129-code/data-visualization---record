Aim:
To study various data visualization tools and display the installed versions of commonly used Python
visualization libraries.

Algorithm:

 Import the required visualization libraries.

 Display the version of each library.

 Execute the program.

 Observe the output.


```python
 import matplotlib
 import seaborn
 import plotly
 import pandas
 print("Matplotlib:", matplotlib.__version__)
 print("Seaborn:", seaborn.__version__)
 print("Plotly:", plotly.__version__)
 print("Pandas:", pandas.__version__)
```

```python

import matplotlib
import matplotlib.pyplot as plt

print("Matplotlib version:", matplotlib.__version__)

subjects = ["Python", "Java", "R", "Tableau", "SQL"]
marks = [85, 78, 82, 90, 88]

plt.bar(subjects, marks, color="skyblue", edgecolor="black")
plt.title("Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
```

Result:

Thus, the versions of the commonly used Python data visualization libraries were displayed successfully.
